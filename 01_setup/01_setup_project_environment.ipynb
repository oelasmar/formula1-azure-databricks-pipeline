# Set-up project environment for Formula 1 Project
1. Create External Location
2. Create Catalog formula1
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in the landing schema

### Access Cloud Storage

```sql
%fs ls 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/landing' 
```

### Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dbcourseextdl9999_formula1_inc
    URL 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'External location for the formula1-inc container'
```

### Create Catalog formula1

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1_inc
   MANAGED LOCATION 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for the formula1 project' ;
```

### Create schemas --> landing | bronze | silver | gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1_inc.landing;

CREATE SCHEMA IF NOT EXISTS formula1_inc.bronze
    MANAGED LOCATION 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS formula1_inc.silver
    MANAGED LOCATION 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS formula1_inc.gold
    MANAGED LOCATION 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/gold';    
```

```sql
USE CATALOG formula1_inc;
```

```sql
SHOW SCHEMAS;
```

### Create Volume Files

```sql
CREATE EXTERNAL VOLUME formula1_inc.landing.files
LOCATION 'abfss://formula1-inc@dbcourseextdl9999.dfs.core.windows.net/landing';
```
