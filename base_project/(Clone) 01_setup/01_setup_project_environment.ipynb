# Set-up project environment for Formula 1 Project
1. Create External Location
2. Create Catalog formula1
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in the landing schema

### Access Cloud Storage

```sql
%fs ls 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/landing'
```

### Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dbcourseextdl9999_formula1
    URL 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'External location for the formula1 container'
```

### Create Catalog formula1

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
   MANAGED LOCATION 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/' 
   COMMENT 'This is the main catalog for the formula1 project' ;
```

### Create schemas --> landing | bronze | silver | gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;

CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/gold';    
```

```sql
SELECT current_catalog();
```

```sql
USE CATALOG formula1;
```

```sql
SHOW SCHEMAS;
```

### Create Volume Files

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@dbcourseextdl9999.dfs.core.windows.net/landing';
```
